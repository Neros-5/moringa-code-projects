# Writing Data to a CSV File

## 1. Import the CSV Module

```python
import csv
```
* `csv` is a built-in Python module for working with CSV files.
* No installation is required.

## 2. Store the Data
```python
students = [
    ['Mary', 30, 'Nakuru'],
    ['Job', 27, 'Mombasa'],
    ['Bob', 34, 'Meru']
]
```
* The outer list contains multiple rows.
* Each inner list represents one row.
* The values inside each row represent columns.

## 3. Open/Create the CSV File
```python
with open('students.csv', 'w') as file:
```
* `open()` opens a file or creates it if it doesn't exist.
* `'students.csv'` is the filename.
* `'w'` means **write mode**.
* `as file` gives the opened file a variable name.
* `with` automatically closes the file when finished.

### Common file modes
* `'r'` → Read
* `'w'` → Write/overwrite
* `'a'` → Append

## 4. Create the CSV Writer
```python
writer = csv.writer(file)
```
* Creates a CSV writer connected to the opened file.
* `writer` is then used to write data into the CSV file.

## 5. Write One Row
```python
writer.writerow(['name', 'age', 'town'])
```
* `writerow()` writes **one row**.
* Here it is used to create the column headings.

Result:
```text
name,age,town
```

## 6. Write Multiple Rows
```python
writer.writerows(students)
```
* `writerows()` writes **multiple rows**.
* Each inner list becomes one row in the CSV.

The resulting file:
```text
name,age,town
Mary,30,Nakuru
Job,27,Mombasa
Bob,34,Meru
```

## Overall Process
```text
Python list
     ↓
Open/Create CSV file
     ↓
Create CSV writer
     ↓
writerow()  → Write one row
     ↓
writerows() → Write multiple rows
     ↓
CSV file
```

### Key Difference
```python
writer.writerow(data)    # One row
writer.writerows(data)   # Multiple rows
```


````python
import csv

student_1 = [
    ['Mary', 30, 'Nakuru'],
    ['Job', 27, 'Mombasa'],
    ['Bob', 34, 'Meru']
]

print(student_1[1]) # Simple test to check if the formating is correct

with open ('student_1.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name', 'age', 'town']) # header
    writer.writerows(student_1) # rows of Information
    
````

### Adding a new Row to the csv Data
This is done by using `'a'` which means *Append*. With the same structure as above, you initialize
the file, opening it using 'a' to just add one or more entries. The rest follows.

The Difference Between to Write and to Append, is that *Write* overrides the existing contents, while *Append* just adds a new Row of data at the end of the csv file without deleting/changing everything.

````python
# Add a student using 'Append'
student_2 = ['Maxmila', 33, 'Marigat']
with open('student_1.csv', 'a', newline="") as f:
    writer = csv.writer(f)
    writer.writerow(student_2)
````

### Reading csv contents
This is done by using `r` which means *Read*. Again following the same conventions, read mode is used it when you want Python to retrieve information from an existing file, rather than change the file.
The `.reader()` funtion returns the row values as lists, while the `.DictReader()` function returns the row values as dictionaries

````python
with open("student_1.csv", "r") as file:
    reader = csv.reader(file) # this gives the contents as lists

    for x in reader: 
        print(x)
````

````python
with open("student_1.csv", "r") as file:
    reader = csv.DictReader(file) # this gives the contents as dictionaries.

    for x in reader: 
            print(x)
````
