# Working with JSON in Python

## What is JSON?
**JSON (JavaScript Object Notation)** is a data format used to store and exchange structured data.
It commonly uses **key-value pairs**:
```json
{
    "name": "Betty",
    "age": 40,
    "height": 5.9
}
```
JSON is commonly used with APIs, web applications, and data storage.

## JSON vs CSV
* **CSV** → best for tabular data (rows and columns).
* **JSON** → best for structured or nested data.
* CSV commonly works like a table.
* JSON commonly works like Python dictionaries and lists.

Example CSV:
```text
name,age,town
Mary,30,Nakuru
Job,27,Mombasa
```

Example JSON:
```json
{
    "name": "Mary",
    "age": 30,
    "town": "Nakuru"
}
```

## Creating a JSON File
```python
import json

people = {
    'name': 'Betty',
    'age': 40,
    'height': 5.9,
    'Gender': 'Female'
}

with open('people.json', 'w') as file:
    json.dump(people, file)
```

### Explanation
* `import json` → imports Python's built-in JSON module.
* `people` → Python dictionary containing the data.
* `open('people.json', 'w')` → creates/opens the JSON file in write mode.
* `as file` → gives the opened file the name `file`.
* `json.dump(people, file)` → converts the Python data into JSON format and writes it to the file.

### Basic Process
```text
Python Dictionary
       ↓
   json.dump()
       ↓
    JSON Data
       ↓
   people.json
```

### Important
Python dictionaries commonly use single quotes:
```python
{'name': 'Betty'}
```

JSON uses double quotes:
```json
{"name": "Betty"}
```

`json.dump()` handles this conversion automatically.


````python
import json

people_1 = {
    'name' : 'Betty',
    'age' : 40,
    'height' : 5.9,
    'Gender' : 'Female'
}

with open('people.json', 'w') as file:
    json.dump(people_1, file)
````

# Python File Modes
File modes tell Python **what we want to do with a file** when using `open()`.
```python
with open('people.json', 'r') as file:
    data = json.load(file)
```

### Common File Modes
| Mode  | Meaning | Purpose                                                 |
| ----- | ------- | ------------------------------------------------------- |
| `'r'` | Read    | Reads an existing file                                  |
| `'w'` | Write   | Creates a file or replaces existing contents            |
| `'a'` | Append  | Adds new content to the end of a file                   |
| `'x'` | Create  | Creates a new file; gives an error if it already exists |

### `'r'` — Read
Used when we want to **read** data from a file.
```python
with open('people.json', 'r') as file:
    data = json.load(file)
```

`'r'` only controls how the file is opened.
`json.load()` reads the JSON data and converts it into Python data.

### `'w'` — Write
Used when we want to **write** data to a file.
```python
with open('people.json', 'w') as file:
    json.dump(people_1, file)
```

⚠️ If the file already contains data, `'w'` **replaces the existing contents**.

### `'a'` — Append
Used to **add data to the end** of an existing file without removing its current contents.
```python
with open('people.txt', 'a') as file:
    file.write('Betty\n')
```

### `'x'` — Create
Used to **create a new file**.
```python
with open('people.json', 'x') as file:
    ...
```

If the file already exists, Python gives an error.

### Quick Reminder
```text
'r' → Read
'w' → Write / Replace
'a' → Add to the end
'x' → Create only
```

### JSON Example
Reading:
```text
people.json
    ↓
'r' → Open for reading
    ↓
json.load()
    ↓
Python dictionary
```

Writing:
```text
Python dictionary
    ↓
json.dump()
    ↓
'w' → Open for writing
    ↓
people.json
```

**Remember:** The file mode controls **how the file is opened**, while `json.load()` and `json.dump()` handle the JSON conversion.


````python
# displaying contents. Conventions for using loops and such for dictionaries apply
with open('people.json', 'r') as file: 
    data = json.load(file)

print(data['age'])
````

# Appending Data to JSON

### `'a'` — Append Mode

The `'a'` mode means **append data to the end of a file**.

```python
with open('people.json', 'a') as file:
    ...
```

However, `'a'` does **not** understand JSON. It simply puts whatever we write at the end of the file.

For example, if a JSON file contains:

```json
[
    {
        "name": "Jackson",
        "age": 35
    }
]
```

Using `'a'` to directly add another dictionary can produce:

```json
[
    {
        "name": "Jackson",
        "age": 35
    }
]
{
    "name": "Sophia",
    "age": 20
}
```

This is **not valid JSON**, because there are now two separate JSON structures in the same file.

### Single JSON Object vs Multiple Objects

A JSON file can contain a **single object**:

```json
{
    "name": "Betty",
    "age": 40
}
```

When loaded with `json.load()`, this becomes a **Python dictionary**.

A dictionary is **mutable**, meaning its existing entries can be changed or new entries can be added:

```python
people['age'] = 41
people['height'] = 5.9
```

However, a dictionary does not use `.append()` or `.extend()` because those are **list methods**.

To store multiple separate dictionaries, the usual structure is a **JSON array containing objects**:

```json
[
    {
        "name": "Betty",
        "age": 40
    },
    {
        "name": "Jackson",
        "age": 35
    }
]
```

This becomes a **Python list containing dictionaries**:

```text
List
 ├── Dictionary
 └── Dictionary
```

The list can then use `.append()` or `.extend()`.

### Adding a Dictionary to Normal JSON

For a JSON file containing a **list of dictionaries**, use:

**Read → Modify → Write**

```python
import json

with open('people.json', 'r') as file:
    people = json.load(file)

people.append({
    'name': 'Sophia',
    'age': 20
})

with open('people.json', 'w') as file:
    json.dump(people, file, indent=4)
```

### Adding Multiple Dictionaries

If you already have multiple dictionaries inside another list, use `.extend()`:

```python
people.extend(people_2)
```

`.append()` adds **one item**:

```python
people.append(dictionary)
```

`.extend()` adds **multiple items from another list**:

```python
people.extend(people_2)
```

### How It Works

```text
people.json
     ↓
'r' → Read the existing JSON
     ↓
json.load()
     ↓
Python list
     ↓
.append() / .extend()
     ↓
Add new dictionary/dictionaries
     ↓
'w' → Write the updated data
     ↓
people.json
```

### Important Difference

`'a'` and `.append()` are **not the same thing**.

```python
'a'
```

→ File mode that adds data to the **end of a file**.

```python
people.append(...)
```

→ Python list method that adds an **item to a list**.

### Remember

```text
'a' → Append to a file
.append() → Add one item to a Python list
.extend() → Add multiple items to a Python list

Single JSON object:
{} → Python dictionary → Can be modified

Multiple JSON objects:
[] → Python list → Can use .append() / .extend()

Normal JSON modification:
Read → Modify → Write
```

**JSON itself is not considered "immutable" here.** The important difference is the **data structure**: `{}` represents an object/dictionary, while `[]` represents an array/list.


````python
# adding new data into the json file.
people_2 = [
    {
        "name": "Betty",
        "age": 40,
        "height": 5.9,
        "Gender": "Female"
    },
    {
        "name": "Jackson",
        "age": 35,
        "height": 6.1,
        "Gender": "Male"
    },
    {
        "name": "Sophia",
        "age": 20,
        "height": 5.7,
        "Gender": "Female"
    }
]

with open('people.json', 'r') as file:
    person = json.load(file)
    people = [person]

people.extend(people_2)

with open('people.json', 'w') as file:
    json.dump(people, file, indent=4)
````
