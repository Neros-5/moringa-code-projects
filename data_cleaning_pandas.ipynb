# DATA CLEANING WITH PANDAS
**What is Data Cleaning?**</br>Data cleaning is the process of identifying and addressing problems that make data unreliable/difficult to analyse. Common data cleaning problems include:
- Missing values
- Duplicate values
- Inconsistent Text
- Incorrect Data Types
- Invalid Values

An outlier which is an unusually high or low value compared to the observation is not automatically an error. For example,if the patient wait time is normally between 10 mins to 15 mins but there is a recorded value of a 90 mins wait time,this is clearly an outlier but that does not mean its erroneous.We will end up deciding whether to drop or retain the value\outlier.However, a negative age value violates the meaning of that column denoting that its erroneous.

**Data Cleaning Process**
- Preserve the raw Data and create a working copy.
- Inspect the Data cleaning issues: missing values, duplicate values, erroneous values, column inconsistencies, null values e.t.c
- Deal with the issues example dropping null values, imputations, introduce uniformity e.t.c
- Save the cleaned copy.
