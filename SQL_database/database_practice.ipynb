```python
import sqlite3
import pandas as pd

moringa_db = sqlite3.connect("moringa.db")

cur = moringa_db.cursor()

create_tbl_query = """
    CREATE TABLE IF NOT EXISTS students (
        student_id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        course TEXT NOT NULL,
        email TEXT UNIQUE NOT NULL,
        gender TEXT NOT NULL,
        bmi REAL NOT NULL
    );
"""

cur.execute(create_tbl_query)

moringa_db.commit()
```
