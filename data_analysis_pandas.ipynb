## Introducton to Data Analysis with Python
**What is Data Analysis?**

Its the process of examining Data to find patterns, answer questions and make decisions. For example a number becomes useful when we understand what it measures, a waiting time of 20 means nothing until we know the unit is in minutes and thus the record describes one clinic visit. 

Data Science is a field that uses data, statistics, domain knowledge and programming to investigate questions and come up with relevant solutions. Data analysis thus becomes a part of this work.

There are four common types of analysis:
- ***Descriptive Analysis:*** Descriptives tell us what happened, for example how many patient visits were recorded per clinic.
- ***Diagnostics Analysis:*** Tells us why it happened, for example busy patient arrival periods might result to longer wait times.
- ***Predictive Analysis:*** Tells us what might happen, for example how many patient visits might be expected next week.
- ***Prescriptive Analysis:*** Tells us what should be done or what action should be taken, for example which staffing plan could reduce patient waiting times.



**Analysis WorkFLow**
- State a question.
- Get/Obtain/Load the Data.
- Inspecting the Data: Checking for null values, duplicate values, outliers/extreme values, Data types, Data size(how many rows vs columns) and erroneous values e.t.c
- Clean the Data.
- Analyze and visualize.
- Interpret and communicate the results.


**Pandas**

## Pandas Series and DataFrames

**Series**</br>
This is a one dimensional labelled collection of values. Think of it as a column together with its rows.

```python
import pandas as pd # pd is an alias

# Pandas Series
age = pd.Series([22,34,27,44,19], name = 'age') # 'name= ' gives the series a descriptive name e.g age, minutes etc
age

```

```python
#create a series that stores temperature. 
temperature = pd.Series([12,24,45,8,40], index=['Monday','Tuesday','Wednesday','Thursday','Friday'], name='temperature')
temperature
```

**DataFrames**</br>Its a two dimensional labelled table with rows and columns i.e like an Excel sheet which is a collection of multiple series (has multiple rows and multiple columns).For example data that has name, age and county.

```python
#create a dictionary.
students = {
    'name': ['John','Mary','Peter'],
    'age' : [22, 34, 27],
    'county': ['Nairobi','Mombasa','Kisumu']
    }

print(students)

#turn it into a DataFrame
df = pd.DataFrame(students)
df
```
